#### US Population by Year (1950–2020)

The following data shows the estimated population of the United States for each year from 1950 to 2020. This information is useful for analyzing demographic trends and understanding historical growth patterns.

*Source: [U.S. Census Bureau - Population Estimates](https://www.census.gov/data/tables/time-series/demo/popest/2020s-national-total.html)*

```python
us_population = {
    1950: 152271417, 1951: 154877889, 1952: 157552740, 1953: 160184192, 1954: 163025854,
    1955: 165931202, 1956: 168903031, 1957: 171984130, 1958: 174881904, 1959: 177829628,
    1960: 180671158, 1961: 183691481, 1962: 186537737, 1963: 189242789, 1964: 191888791,
    1965: 194302963, 1966: 196560338, 1967: 198712056, 1968: 200706052, 1969: 202676946,
    1970: 205052174, 1971: 207660677, 1972: 209896021, 1973: 211908788, 1974: 213853928,
    1975: 215973199, 1976: 218035164, 1977: 220239425, 1978: 222584545, 1979: 225055487,
    1980: 227224681, 1981: 229465714, 1982: 231664458, 1983: 233792014, 1984: 235824902,
    1985: 237923795, 1986: 240132887, 1987: 242288918, 1988: 244499004, 1989: 246819230,
    1990: 249623822, 1991: 252981685, 1992: 256514224, 1993: 259919889, 1994: 263125821,
    1995: 266278403, 1996: 269394284, 1997: 272646925, 1998: 275854104, 1999: 279040168,
    2000: 282162411, 2001: 284968955, 2002: 287625193, 2003: 290107933, 2004: 292805298,
    2005: 295516599, 2006: 298379912, 2007: 301231207, 2008: 304093966, 2009: 306771529,
    2010: 309326085, 2011: 311580009, 2012: 313874218, 2013: 316057727, 2014: 318386329,
    2015: 320738994, 2016: 323071755, 2017: 325122128, 2018: 327167439, 2019: 328239523,
    2020: 331449281
}
```

```python
# Convert the dictionary to a DataFrame
years = np.array(list(us_population.keys()))
populations = np.array(list(us_population.values()))
df = pd.DataFrame({'Year': years, 'Population': populations})

# Set projection period (modifiable)
projection_years = 5

# Linear Projection using numpy polyfit (degree 1)
linear_coeffs = np.polyfit(df['Year'], df['Population'], 1)
linear_model = np.poly1d(linear_coeffs)
future_years = np.arange(df['Year'].max() + 1, df['Year'].max() + projection_years + 1)
linear_projection = linear_model(future_years)

# Comparative Growth Rate Projection (average annual growth rate)
growth_rates = df['Population'].pct_change().dropna()
avg_growth_rate = statistics.mean(growth_rates)
comp_growth_projection = [df['Population'].iloc[-1]]
for _ in range(projection_years):
    comp_growth_projection.append(comp_growth_projection[-1] * (1 + avg_growth_rate))
comp_growth_projection = comp_growth_projection[1:]

# Advanced Projection: Simple Exponential Smoothing (custom implementation)
alpha = 0.8  # smoothing factor, can be tuned
ses_projection = [df['Population'].iloc[-1]]
for _ in range(projection_years):
    next_val = alpha * ses_projection[-1] + (1 - alpha) * df['Population'].iloc[-1]
    ses_projection.append(next_val)
ses_projection = ses_projection[1:]

# Plotting
plt.figure(figsize=(12, 6))
plt.plot(df['Year'], df['Population'], label='Historical Data', marker='o')
plt.plot(future_years, linear_projection, label='Linear Projection', linestyle='--')
plt.plot(future_years, comp_growth_projection, label='Comparative Growth Rate', linestyle='-.')
plt.plot(future_years, ses_projection, label="Simple Exponential Smoothing", linestyle=':')
plt.xlabel('Year')
plt.ylabel('Population')
plt.title('US Population Projection')
plt.legend()
plt.grid(True)
plt.show()
```
